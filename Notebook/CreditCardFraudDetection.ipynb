Preparing Data for Machine Learning:

The dataset is divided into input features (X) and the target variable (y).

X contains the transaction-related features used by the machine learning model.

y contains the `Class` column, which indicates whether the transaction is
genuine or fraudulent.

```python
X = df.drop('Class', axis=1)
y = df['Class']

print("Features (X):", X.shape)
print("Target (y):", y.shape)
```

Splitting the Dataset:

The dataset is divided into training and testing sets.

The training set is used to train the machine learning model,
while the testing set is used to evaluate the model on unseen data.

An 80:20 ratio is used. Stratification is applied to maintain
the proportion of genuine and fraudulent transactions in both sets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

Class Distribution After Splitting

Since fraudulent transactions are much fewer than genuine transactions,
we check the class distribution in the training and testing datasets.

Stratified splitting was used to maintain a similar proportion of
genuine and fraudulent transactions in both datasets.

```python
print("Training data class distribution:")
print(y_train.value_counts())

print("\nTesting data class distribution:")
print(y_test.value_counts())
```

Feature Scaling

Feature scaling is performed to bring features with different numerical
ranges to a comparable scale.

In this project, the `Time` and `Amount` features are standardized using
StandardScaler.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train[['Time', 'Amount']] = scaler.fit_transform(
    X_train[['Time', 'Amount']]
)

X_test[['Time', 'Amount']] = scaler.transform(
    X_test[['Time', 'Amount']]
)
```

Logistic Regression

Logistic Regression is used to classify transactions as genuine or fraudulent.

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

Making Predictions

The trained model is used to predict whether the test transactions are genuine or fraudulent.

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
```

Confusion Matrix

The confusion matrix compares the actual transaction classes with the model's predictions.

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

Confusion Matrix Visualization

The confusion matrix is displayed as a heatmap to make the model's predictions easier to understand.

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Genuine', 'Fraud'],
    yticklabels=['Genuine', 'Fraud']
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

Evaluation Metrics

Precision, Recall, F1-score, and Accuracy are used to evaluate the model.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-Score :", f1)
```

Classification Report

The classification report shows the precision, recall, and F1-score for genuine and fraudulent transactions.

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

Improved Logistic Regression

Class weighting is used to give more importance to the minority fraud class during training.

```python
from sklearn.linear_model import LogisticRegression

balanced_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

balanced_model.fit(X_train, y_train)

y_pred_balanced = balanced_model.predict(X_test)

print("Balanced Logistic Regression trained successfully!")
```

Evaluating the Improved Model

The improved model is evaluated using precision, recall, and F1-score.

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_balanced))
```

Comparing the Models

The original and balanced Logistic Regression models are compared using precision, recall, and F1-score for fraud detection.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

original_precision = precision_score(y_test, y_pred)
original_recall = recall_score(y_test, y_pred)
original_f1 = f1_score(y_test, y_pred)

balanced_precision = precision_score(y_test, y_pred_balanced)
balanced_recall = recall_score(y_test, y_pred_balanced)
balanced_f1 = f1_score(y_test, y_pred_balanced)

comparison = pd.DataFrame({
    'Model': ['Original Logistic Regression',
              'Balanced Logistic Regression'],
    'Precision': [original_precision, balanced_precision],
    'Recall': [original_recall, balanced_recall],
    'F1-Score': [original_f1, balanced_f1]
})

comparison
```

Random Forest

Random Forest is an ensemble machine learning algorithm that combines multiple decision trees to make predictions.

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    class_weight='balanced'
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest model trained successfully!")
```

Random Forest Evaluation

The Random Forest model is evaluated using precision, recall, and F1-score.

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_rf))
```

Model Comparison

The performance of the three models is compared using precision, recall, and F1-score for fraud detection.

```python
comparison = pd.DataFrame({
    'Model': [
        'Original Logistic Regression',
        'Balanced Logistic Regression',
        'Random Forest'
    ],
    'Precision': [
        original_precision,
        balanced_precision,
        precision_score(y_test, y_pred_rf)
    ],
    'Recall': [
        original_recall,
        balanced_recall,
        recall_score(y_test, y_pred_rf)
    ],
    'F1-Score': [
        original_f1,
        balanced_f1,
        f1_score(y_test, y_pred_rf)
    ]
})

comparison
```

Feature Importance

Feature importance shows which features contribute most to the Random Forest model's predictions.

```python
feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

feature_importance.head(10)
```

Top 10 Important Features

The graph shows the features that were most important to the Random Forest model.

```python
plt.figure(figsize=(8, 5))

top_features = feature_importance.head(10)

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title('Top 10 Important Features')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
```

ROC-AUC Evaluation

ROC-AUC measures how well the model distinguishes between genuine and fraudulent transactions.

```python
from sklearn.metrics import roc_auc_score

y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob_rf)

print("Random Forest ROC-AUC:", roc_auc)
```

Precision-Recall Curve

The Precision-Recall curve shows the trade-off between precision and recall when detecting fraudulent transactions.

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_prob_rf
)

plt.figure(figsize=(8, 5))

plt.plot(recall, precision)

plt.title('Precision-Recall Curve')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.grid()

plt.show()
```

Selecting the Final Model

The models are compared using precision, recall, and F1-score.
The model with the most suitable fraud detection performance will be selected as the final model.

```python
comparison.sort_values(
    by='F1-Score',
    ascending=False
)
```

Saving the Final Model

The Random Forest model is selected as the final model and saved for future predictions.

```python
import joblib

joblib.dump(rf_model, "../rf_fraud_detection_model.pkl")

print("Final Random Forest model saved successfully!")
```

Fraud Prediction Function

A prediction function is created to classify a transaction as genuine or fraudulent using the trained Random Forest model.

```python
def predict_fraud(transaction):
    prediction = rf_model.predict(transaction)

    if prediction[0] == 1:
        return "Fraudulent Transaction"
    else:
        return "Genuine Transaction"
```

Testing the Prediction Function

A sample transaction from the test dataset is used to verify that the prediction function works correctly.

```python
sample_transaction = X_test.iloc[[0]]

result = predict_fraud(sample_transaction)

print("Prediction:", result)
```

Fraud Probability

The model's fraud probability is calculated for a sample transaction.

```python
sample_transaction = X_test.iloc[[0]]

fraud_probability = rf_model.predict_proba(sample_transaction)[0][1]

prediction = rf_model.predict(sample_transaction)[0]

print("Fraud Probability:", round(fraud_probability * 100, 2), "%")

if prediction == 1:
    print("Prediction: Fraudulent Transaction")
else:
    print("Prediction: Genuine Transaction")
```
